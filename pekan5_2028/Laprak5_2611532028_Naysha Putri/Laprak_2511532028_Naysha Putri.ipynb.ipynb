## 2.1 Dasar Teori
## DASAR TEORI

### 2.1.1 Perulangan `for`

Perulangan `for` digunakan untuk menjalankan suatu perintah secara berulang berdasarkan nilai yang telah ditentukan. Pada materi praktikum pekan ke-5, perulangan `for` digunakan untuk menghasilkan beberapa baris output secara teratur. Salah satu bentuk dasarnya adalah menentukan batas perulangan menggunakan `range()`.

Contoh bentuk perulangan:

```python
for count in range(1, batas + 1):
    ...
```

Perulangan tersebut akan menjalankan proses mulai dari nilai 1 sampai nilai `batas`. Materi praktikum juga menggunakan konsep ini sebagai **outer loop**, yaitu perulangan yang mengatur setiap baris pada suatu pola.

### 2.1.2 Outer Loop dan Inner Loop

Dalam pembuatan pola yang kompleks, digunakan dua tingkat perulangan, yaitu **outer loop** dan **inner loop**. Outer loop digunakan untuk mengatur jumlah baris yang akan dibuat, sedangkan inner loop digunakan untuk menghasilkan bagian-bagian pola pada setiap baris.

Materi praktikum menjelaskan bahwa untuk membuat beberapa baris pola, diperlukan outer loop yang berjalan sebanyak jumlah baris dan inner loop yang mengulang karakter sesuai pola pada setiap baris.

Konsep tersebut dapat digambarkan sebagai:

```python
for baris_2028 in range(1, batas_2028 + 1):
    for karakter_2028 in range(...):
        print(..., end="")
    print()
```

Pada program Jam Pasir Kristal Palindromik, konsep ini digunakan untuk mengatur setiap baris dan kemudian mengatur spasi serta angka yang terdapat pada baris tersebut.

### 2.1.3 Nested `for` Loop

Nested `for` loop adalah perulangan `for` yang berada di dalam perulangan `for` lainnya. Nested loop digunakan untuk menghasilkan pola yang memiliki beberapa bagian dalam setiap baris.

Pada materi PDF, contoh nested loop digunakan untuk menghasilkan pola:

```text
....1
...2
..3
.4
5
```

Outer loop digunakan untuk menentukan nomor baris, sedangkan inner loop digunakan untuk mencetak jumlah titik yang berbeda pada setiap baris.

Pada tugas ini, konsep yang sama diterapkan untuk menghasilkan spasi, angka menurun, poros `<*>`, dan angka menaik pada setiap baris jam pasir.

### 2.1.4 Penggunaan `range()`

Fungsi `range()` digunakan untuk menentukan urutan nilai yang akan diproses oleh perulangan `for`. Penggunaan `range()` dapat disesuaikan dengan kebutuhan pola, baik untuk perulangan menaik maupun menurun.

Contohnya:

```python
for angka_2028 in range(1, 5):
    print(angka_2028)
```

akan menghasilkan:

```text
1
2
3
4
```

Sedangkan untuk menghasilkan urutan menurun dapat digunakan nilai langkah negatif:

```python
for angka_2028 in range(4, 0, -1):
    print(angka_2028)
```

yang menghasilkan:

```text
4
3
2
1
```

Konsep perulangan menaik dan menurun tersebut digunakan dalam pembentukan bagian angka pada pola jam pasir.

### 2.1.5 Perulangan untuk Membentuk Pola

Pembuatan pola menggunakan `for` dilakukan dengan menghubungkan jumlah perulangan dengan posisi karakter pada setiap baris. Pada materi praktikum, jumlah karakter yang dicetak dapat ditentukan berdasarkan hubungan antara nomor baris dan batas pola.

Sebagai contoh, pada pola:

```text
....1
...2
..3
.4
5
```

jumlah titik pada setiap baris adalah:

| Baris | Jumlah Titik |
| ----- | -----------: |
| 1     |            4 |
| 2     |            3 |
| 3     |            2 |
| 4     |            1 |
| 5     |            0 |

Materi menunjukkan bahwa jumlah titik tersebut dapat diperoleh dari formula:

```text
-1 × line + batas
```

sehingga ketika `line` bertambah, jumlah titik berkurang satu.

Konsep formula seperti ini digunakan dalam tugas untuk menentukan jumlah spasi pada setiap baris sehingga pola jam pasir dapat terbentuk secara simetris.

### 2.1.6 Analisis Formula pada Perulangan

Dalam membuat pola menggunakan perulangan, formula dapat digunakan untuk menentukan hubungan antara nomor baris dengan nilai yang akan dicetak. Materi praktikum memberikan contoh penggunaan tabel untuk menemukan pola bilangan.

Sebagai contoh, untuk menghasilkan:

```text
2 7 12 17 22
```

setiap kenaikan nilai `count` menyebabkan bilangan bertambah 5. Formula yang diperoleh adalah:

```text
5 × count - 3
```

Materi menggunakan tabel untuk membantu menemukan hubungan tersebut.

Cara yang sama dapat digunakan untuk menganalisis jumlah spasi pada program jam pasir. Dengan mengetahui hubungan antara `N` dan nomor baris, jumlah spasi dapat ditentukan menggunakan formula:

```text
2 × (N - baris)
```

Formula tersebut membuat jarak pola semakin besar ketika baris mendekati bagian tengah jam pasir.

### 2.1.7 Parameter `end=""`

Dalam pembuatan pola, beberapa karakter harus dicetak pada baris yang sama. Oleh karena itu, digunakan parameter `end=""` pada fungsi `print()`.

Contohnya:

```python
print("1", end="")
print("2", end="")
print("3", end="")
```

Hasilnya:

```text
123
```

Penggunaan `end=""` memungkinkan setiap karakter dicetak tanpa langsung berpindah ke baris berikutnya. Setelah seluruh bagian pada satu baris selesai, `print()` tanpa argumen digunakan untuk berpindah ke baris berikutnya. Konsep pencetakan karakter berulang dengan `for` dan `end=""` juga ditunjukkan dalam contoh nested loop pada materi praktikum.

### 2.1.8 Penerapan Nested Loop pada Pola Jam Pasir

Pada tugas praktikum ini, konsep `for` dan nested loop diterapkan untuk membuat **Pola Jam Pasir Kristal Palindromik Berbingkai**. Outer loop digunakan untuk mengatur baris jam pasir, sedangkan inner loop digunakan untuk membentuk bagian-bagian pada setiap baris.

Setiap baris terdiri dari beberapa bagian, yaitu spasi penyeimbang, angka yang tersusun menurun, poros kristal `<*>`, angka yang tersusun menaik, dan spasi penyeimbang. Dengan mengatur jumlah pengulangan setiap bagian menggunakan formula tertentu, pola dapat berubah secara dinamis berdasarkan nilai `N` yang dimasukkan pengguna.

Konsep ini sesuai dengan materi praktikum yang menunjukkan bahwa pola kompleks dapat dibentuk dengan menggabungkan outer loop untuk setiap baris dan inner loop untuk bagian pola di dalam baris tersebut.

Contohnya jika pengguna memasukkan 10, bilangan yang dijumlahkan adalah:

```text
2 + 4 + 6 + 8 + 10 = 30
```

Program ini menunjukkan penerapan perulangan `for` yang digabungkan dengan percabangan `if` untuk melakukan penyaringan data berdasarkan kondisi tertentu.

### 9. Program `tugas5`

Program kesembilan digunakan untuk membuat **pola segitiga menggunakan karakter `*`**. Program meminta pengguna memasukkan tinggi segitiga melalui variabel `tinggi_2028`.

Perulangan luar:

```python
for i_2028 in range(1, tinggi_2028 + 1):
```

digunakan untuk menentukan jumlah baris segitiga.

Kemudian terdapat dua perulangan dalam. Perulangan pertama digunakan untuk mencetak spasi:

```python
for j_2028 in range(tinggi_2028 - i_2028):
    print(" ", end="")
```

Jumlah spasi semakin berkurang ketika nomor baris bertambah.

Perulangan kedua digunakan untuk mencetak karakter `*`:

```python
for k_2028 in range(i_2028):
    print("*", end=" ")
```

Jumlah `*` bertambah sesuai dengan nomor baris. Setelah seluruh bagian pada satu baris selesai dicetak, `print()` digunakan untuk berpindah ke baris berikutnya.

Contoh jika tinggi = 5:

```text
    *
   * *
  * * *
 * * * *
* * * * *
```

Program ini merupakan penerapan nested loop untuk mengatur **spasi dan karakter secara bersamaan**, sehingga membentuk pola segitiga.
